#-------List (Creation, Modification and Access):---------

```python
age_list = [24, 25, 26, 27, 28]
print(age_list)
```

```python
name_list = ["Sudhishma", "Naveen", "Mounika", "Varadh", "Shloka"]
print(name_list)
```

#----------List Operations / Modifications-----------

```python
name_list.append("Yazhini")
print(name_list)
```

```python
age_list.insert(2, 30)
print(age_list)
```

```python
name_list.remove("Yazhini")
print(name_list)
```

```python
age_list = [24, 25, 30, 26, 27, 28]
popped_element = age_list.pop()
print("Popped element:", popped_element)
print("Updated age_list:", age_list)
```

```python
age_list.extend([29, 30, 26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
max_age = max(age_list)
min_age = min(age_list)
total_sum = sum(age_list)

print("Max age:", max_age)
print("Min age:", min_age)
print("Sum of all ages:", total_sum)
```

#---------------Sets (Operations):---------------------

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
#Since sets in Python only store unique elements and do not maintain order, duplicate items like 'a' and 'i' are automatically discarded.
```

```python
my_set[4] = 's'

print(my_set)
#Python sets do not maintain elements in a specific order or position. Because there is no concept of "4th element" or index 4 in a set,
indexing with square brackets [] is not supported.
#Item Assignment is Disabled: You cannot modify a set element in place by replacing it at a specific position.    
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
set_union = set1.union(set2)
set_intersection = set1.intersection(set2)

print("Union:", set_union)
print("Intersection:", set_intersection)
```
