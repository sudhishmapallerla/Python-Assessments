```python
# numpy array operations
```

```python
# Create the 1D NumPy array for Week 1 temperatures
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
array_shape = temperatures_w1.shape

array_dtype = temperatures_w1.dtype

array_size = temperatures_w1.size



print(f"Array Shape: {array_shape}")
print(f"Data Type: {array_dtype}")
print(f"Number of Elements: {array_size}")
```

```python
# Convert the array to Fahrenheit
temperatures_w1_f = (temperatures_w1 * 9/5) + 32

print(temperatures_w1_f)
```

```python
# Calculate statistics for Celsius
max_c = temperatures_w1.max()
min_c = temperatures_w1.min()
mean_c = temperatures_w1.mean()


print(f"Celsius    -> Max: {max_c}°C | Min: {min_c}°C | Mean: {mean_c:.2f}°C")
```

```python
#Array Slicing and Indexing:
```

```python
# Slice the first three elements 
first_three_days = temperatures_w1[:3]

# Slice the last two elements
weekend_temps = temperatures_w1[-2:]

# Slice the middle three elements
middle_three_days = temperatures_w1[2:5]



print(first_three_days)
print(weekend_temps)
print(middle_three_days)
```

```python
#Create a 2D Array:
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   
])


print(temperatures)
```

```python
#Inspect and Slice the 2D Array:
```

```python
print(f"Array Shape: {temperatures.shape}")
print(f"Data Type: {temperatures.dtype}")
print(f"Total Number of Elements: {temperatures.size}")
```

```python
# 1. Extract the temperatures for each week individually
week1_temps = temperatures[0, :]
week2_temps = temperatures[1, :]

# 2. Extract the weekends (last two days) for both weeks simultaneously
# ':' selects all rows (both weeks), and '-2:' selects the last two columns
both_weekends = temperatures[:, -2:]

# Print the results
print("Week 1 Temperatures:", week1_temps)
print("Week 2 Temperatures:", week2_temps)
print("Weekend Temperatures (Both Weeks):\n", both_weekends)
```

```python
#Creating Pandas Series:
```

```python
import pandas as pd

marks_data = [95, 92, 89, 85, 80]
ranks_index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']

marks = pd.Series(data=marks_data, index=ranks_index)

print(marks)
```

```python
#Indexing and Slicing:
```

```python
first_rank_mark = marks.iloc[0]

print(f"The mark of the 1st rank student is: {first_rank_mark}")
```

```python
top_three_marks = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print(top_three_marks)
```

```python
third_rank_mark = marks.iloc[2]

print(f"The mark of the 3rd rank student is: {third_rank_mark}")
```

```python
# Create and apply the boolean mask
high_ranks = marks[marks > 90]

print(high_ranks)
```

```python
#3. Manipulating Series:
```

```python
marks.loc['Rank1'] = 100
# Drop the last rank from the Series
marks = marks.drop('Rank5')

print(marks)
```

```python
cgpa = marks / 10

print(cgpa)
```

```python
#Pandas DataFrame:

#1. Creating Pandas DataFrame:
```

```python
# Create the transactions DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 
                        'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 
               'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
#Data Exploration:
```

```python
print(transactions.head())
print(transactions.tail())
print("Shape:", transactions.shape)
print(f"Column Names: {list(transactions.columns)}")
print(transactions.dtypes)
```

```python
# Select and display specific columns
product_amounts = transactions[['ProductCategory', 'Amount']]

print(product_amounts)
```

```python
last_three_cols = transactions.iloc[:, -3:]

print(last_three_cols)
```

```python
# Filter rows where Region is 'North' AND Amount is greater than 200
filtered_df = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]

print(filtered_df)
```

```python
# Get the value counts for the ProductCategory column
category_counts = transactions['ProductCategory'].value_counts()

print(category_counts)
```

```python
# Find the unique values in the Region column
unique_regions = transactions['Region'].unique()

print(unique_regions)
```

```python

# Group by Region and calculate the mean of the Amount column
region_means = transactions.groupby('Region')['Amount'].mean()

print(region_means)
```

```python
#3. Manipulating the DataFrame:
```

```python
# Update Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions.head())
```

```python
# Add the 'Discount' column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
# Remove the row where TransactionID is 109
transactions = transactions[transactions['TransactionID'] != 109]

print(transactions)
```

```python
# Drop the 'Discount' column
transactions = transactions.drop('Discount', axis=1)

print(transactions)
```
