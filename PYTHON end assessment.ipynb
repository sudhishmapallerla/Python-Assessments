Project Title: Survey Feedback Analyzer

```python
feedback_data = {
    "response_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "customer_name": [
        "Alice", "Bob", "Charlie", "David", "Eve",
        "Frank", "Grace", "Hannah", "Ian", "Jack"
    ],
    "category": [
        "Product", "Service", "Product", "Pricing", "Service",
        "Product", "Pricing", "Service", "Product", "Service"
    ],
    "rating": [5, 1, 4, 2, 3, 5, 2, 4, 1, 5],
    "comments": [
        "  Great product! Very easy to use and fast. ",
        "Terrible support! Slow response and unhelpful staff.",
        "Good quality overall, but a bit expensive.",
        "Overpriced and bad customer service.",
        "Average experience, could be much better.",
        "EXCELLENT build quality and great speed! ",
        "Way too expensive for the features provided.",
        "Friendly customer support and quick resolution.",
        "Defective item received, very disappointed.",
        "Superb experience! Highly recommended."
    ]
}

# 1. Ensure S_No list exists for preloaded entries (1-10)
if "S_No" not in feedback_data:
    feedback_data["S_No"] = list(range(1, len(feedback_data["customer_name"]) + 1))

# 2. Get number of new records to collect
num_entries = int(input("How many more feedbacks do you want to add? "))

# 3. Collect inputs and append directly to feedback_data dictionary
for i in range(num_entries):
    print(f"\n--- Adding Feedback #{i + 1} ---")
    
    # Auto-increment S_No (starts at 11 if 10 records exist)
    next_s_no = feedback_data["S_No"][-1] + 1 if feedback_data["S_No"] else 1
    new_id = 100 + next_s_no
    
    name = input("Enter Customer Name: ").strip()
    comment = input("Enter Written Feedback: ").strip()
    
    # Input validation for Rating (1-5)
    while True:
        rating_input = input("Enter Rating (1-5): ").strip()
        if rating_input.isdigit() and 1 <= int(rating_input) <= 5:
            rating = int(rating_input)
            break
        print("Invalid entry. Please enter a whole number from 1 to 5.")
        
    category = input("Enter Category (Product/Service/Pricing): ").strip()
    if not category:
        category = "General"

    # Append all attributes into feedback_data dictionary lists
    feedback_data["S_No"].append(next_s_no)
    feedback_data["response_id"].append(new_id)
    feedback_data["customer_name"].append(name)
    feedback_data["category"].append(category)
    feedback_data["rating"].append(rating)
    feedback_data["comments"].append(comment)
    
    print(f"✓ Recorded: S_No {next_s_no} | ID {new_id}")

print(f"\nSuccessfully updated dictionary! Total entries: {len(feedback_data['S_No'])}")
```

#----------Step 3: Text Cleaning 

```python
punctuation_marks = [".", ",", "!", "?"]

cleaned_comments = []

for comment in feedback_data["comments"]:
    cleaned_text = comment
    for char in punctuation_marks:
        cleaned_text = cleaned_text.replace(char, "")
    cleaned_comments.append(cleaned_text)

feedback_data["cleaned_comments"] = cleaned_comments

print("--- Punctuation Removal Sample ---")
for i in range(min(5, len(feedback_data["comments"]))):
    print(f"Original: {feedback_data['comments'][i]}")
    print(f"Cleaned : {feedback_data['cleaned_comments'][i]}\n")
```

```python
space_cleaned_comments = []

source_comments = feedback_data.get("cleaned_comments", feedback_data["comments"])

for comment in source_comments:
    
    normalized_text = " ".join(comment.split())
    space_cleaned_comments.append(normalized_text)

feedback_data["cleaned_comments"] = space_cleaned_comments

print("--- Space Normalization Sample ---")
for i in range(min(5, len(feedback_data["comments"]))):
    print(f"Before: '{source_comments[i]}'")
    print(f"After : '{feedback_data['cleaned_comments'][i]}'\n")
```

```python
# List to store comments with leading/trailing spaces removed
stripped_comments = []

# Source comments from previous cleaning step or original list
source_comments = feedback_data.get("cleaned_comments", feedback_data["comments"])

for comment in source_comments:
    # Remove leading and trailing whitespace
    clean_text = comment.strip()
    stripped_comments.append(clean_text)

# Update the cleaned_comments list in feedback_data
feedback_data["cleaned_comments"] = stripped_comments

# Verification output
print("--- Leading & Trailing Space Removal Sample ---")
for i in range(min(5, len(feedback_data["comments"]))):
    print(f"Original: '{feedback_data['comments'][i]}'")
    print(f"Cleaned : '{feedback_data['cleaned_comments'][i]}'\n")
    
```

```python
# List to store lowercased comments
lowercased_comments = []

# Source comments from previous cleaning steps
source_comments = feedback_data.get("cleaned_comments", feedback_data["comments"])

for comment in source_comments:
    # Convert text to lowercase
    clean_text = comment.lower()
    lowercased_comments.append(clean_text)

# Update feedback_data dictionary
feedback_data["cleaned_comments"] = lowercased_comments

# Verification sample
print("--- Lowercase Conversion Sample ---")
for i in range(min(5, len(feedback_data["comments"]))):
    print(f"Original: '{feedback_data['comments'][i]}'")
    print(f"Cleaned : '{feedback_data['cleaned_comments'][i]}'\n")
```

#--------------- Step 4: Word Count Insights 

```python
def count_word_in_feedbacks(word):
    """Counts total occurrences of a word across all cleaned feedback entries."""
    # Convert search word to lowercase for case-insensitive matching
    target_word = word.strip().lower()

    total_occurrences = 0
    feedback_matches = 0

    # Retrieve cleaned comments list (fallback to raw comments if needed)
    comments_list = feedback_data.get(
        "cleaned_comments", feedback_data["comments"]
    )

    for comment in comments_list:
        # Split comment into individual word tokens
        words = comment.lower().split()
        match_count = words.count(target_word)

        if match_count > 0:
            total_occurrences += match_count
            feedback_matches += 1

    print(
        f"Result: The word '{target_word}' appears {total_occurrences} time(s) across {feedback_matches} feedback entry/entries."
    )
    return total_occurrences


# Example Usage:
count = count_word_in_feedbacks("product")
count_word_in_feedbacks("expensive")
```

```python
def count_word_in_feedbacks(word):
    """Returns how many feedback entries contain the specified word (case-insensitive)."""
    target_word = word.strip().lower()
    matching_count = 0

    # Retrieve cleaned comments list (fallback to raw comments if needed)
    comments_list = feedback_data.get(
        "cleaned_comments", feedback_data["comments"]
    )

    for comment in comments_list:
        # Split into individual word tokens for exact word matching
        words = comment.lower().split()

        # Check if target word exists in this specific feedback entry
        if target_word in words:
            matching_count += 1

    return matching_count


# Example Usage:
search_term = "product"
result = count_word_in_feedbacks(search_term)
print(f"Number of feedbacks containing '{search_term}': {result}")
```

```python
comments = feedback_data.get("cleaned_comments", feedback_data["comments"])

max_words = -1
longest_index = -1

for i, comment in enumerate(comments):
    # Calculate word count by splitting text into words
    word_count = len(comment.split())
    
    if word_count > max_words:
        max_words = word_count
        longest_index = i

print("--- Feedback with the Longest Comment ---")
print(f"S_No: {feedback_data.get('S_No', list(range(1, len(comments) + 1)))[longest_index]}")
print(f"Response ID: {feedback_data['response_id'][longest_index]}")
print(f"Customer Name: {feedback_data['customer_name'][longest_index]}")
print(f"Rating: {feedback_data['rating'][longest_index]}")
print(f"Word Count: {max_words} words")
print(f"Comment: \"{comments[longest_index]}\"")
```

```python
comments = feedback_data.get("cleaned_comments", feedback_data["comments"])

unique_words = set()

for comment in comments:
    # Split text into individual words and add them to the set
    words = comment.lower().split()
    unique_words.update(words)

unique_words_list = sorted(list(unique_words))

print(f"Total Unique Words: {len(unique_words_list)}")
print("List of Unique Words:")
print(unique_words_list)
```

```python
# Zip parallel lists from feedback_data together
zipped_feedbacks = list(
    zip(
        feedback_data["response_id"],
        feedback_data["customer_name"],
        feedback_data["rating"],
        feedback_data.get("cleaned_comments", feedback_data["comments"]),
    )
)

# Sort by rating (3rd item in tuple, index 2) from highest to lowest
sorted_feedbacks = sorted(zipped_feedbacks, key=lambda x: x[2], reverse=True)

# Display sorted results
print(
    f"{'Rank':<6} | {'ID':<6} | {'Customer':<12} | {'Rating':<8} | {'Cleaned Comment'}"
)
print("-" * 75)

for rank, (r_id, name, rating, comment) in enumerate(sorted_feedbacks, start=1):
    print(f"{rank:<6} | {r_id:<6} | {name:<12} | {rating:<8} | {comment}")
```
