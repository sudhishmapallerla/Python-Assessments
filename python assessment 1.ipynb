#---------Strings---#

```python
str1 = "Hello "
name = input("Sudhishma: ")  
result = str1 + name

str3 = ", welcome to Python programming"
final_result = result + str3

print(final_result)
```

```python
text = "Hello Sudhishma, welcome to Python programming"
print(text[0])
print(text[-1])
print(text[:5])
print(text[-11:])
print(text[::-1])
print(text[28:34])
```

```python
strM = "Python beginner tutorial"
print(strM)
print(strM.upper())
print(strM.lower())
print(strM.lower().capitalize())
print(strM.count('t'))
new_str = strM.replace("Python", "Machine Learning")
print(new_str)
```
