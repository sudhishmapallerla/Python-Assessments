#--------------While loop & Control Statements-----------------------

```python
import random

# 1. Set Up the Game
secret_number = random.randint(1, 10)

# 2. Set maximum attempts
attempts = 3

# 3 & 4. Game Loop with continue, break, and else
while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Skip invalid range without deducting an attempt
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue  # continue control statement

    attempts -= 1

    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break  
    elif guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Too high. Try again.")
else:
    
    print("Better luck next time! You ran out of attempts.")
```

#-------------For Loop---------------------

```python
number = int(input("Enter the number for which you want the multiplication table: "))
for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")
```

#----------------Functions:---------------------------

```python
def calculate_bmi(weight, height):
    return weight / (height ** 2)
try:
    user_weight = float(input("Enter your weight in kilograms (kg): "))
    user_height = float(input("Enter your height in meters (m): "))

    if user_weight <= 0 or user_height <= 0:
        print("Please enter positive values greater than zero.")
    else:
       
        calculated_bmi = calculate_bmi(user_weight, user_height)
        print(f"Your calculated BMI is: {calculated_bmi:.2f}")

except ValueError:
    print("Invalid input. Please enter numbers only.")
```
